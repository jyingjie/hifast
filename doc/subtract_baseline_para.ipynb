```python
import numpy as np
import h5py
import matplotlib.pyplot as plt
```

```python
from fast_python.baseline import subtract_baseline
```

```python
#fname= '../G15_drift_5_arcdrift-M03_W-specs_T.hdf5'
fname= 'XXXXXXXXXXXXXXXXX'
```

```python
f= h5py.File(fname,'r')
```

```python
freq= f['freq'][()]

spec= f['T'][0]
```

```python
frange=[1386,1421]

fit_args={}
```

## $\lambda$ 和deg

- $\lambda$ 取值可以从 $10^4$ 到 $10^{13}$
- deg 类似np.polyfit的参数deg， 设为2或3； 不同deg 配合不同 $\lambda$

```python
fit_args['lam']= 1e10
fit_args['deg']= 2

fig,axs= plt.subplots(3,1,figsize=(20,8))
x,ys= subtract_baseline(freq,spec,frange, axs=axs[:2], merge=False,**fit_args)
ax=axs[2]
ax.plot(x,ys,label="After subtracted baseline")
ax.axhline(y=0,color='k',linewidth=1,)
[ax.legend() for ax in axs]
plt.show()
```

```python
fit_args['lam']= 1e5
fit_args['deg']= 2

fig,axs= plt.subplots(3,1,figsize=(20,8))
x,ys= subtract_baseline(freq,spec,frange, axs=axs[:2], merge=False,**fit_args)
ax=axs[2]
ax.plot(x,ys,label="After subtracted baseline")
ax.axhline(y=0,color='k',linewidth=1,)
[ax.legend() for ax in axs]
plt.show()
```

```python
fit_args['lam']= 1e12
fit_args['deg']= 3

fig,axs= plt.subplots(3,1,figsize=(20,8))
x,ys= subtract_baseline(freq,spec,frange, axs=axs[:2], merge=False,**fit_args)
ax=axs[2]
ax.plot(x,ys)
ax.axhline(y=0,color='k',linewidth=1,)
plt.show()
```
